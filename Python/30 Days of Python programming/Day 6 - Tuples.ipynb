$$ Day 5 - Tuples $$

```
Tuples:
- These are the collection of different datatypes which is ordered and unchangable (Immutable).
- Enclosed with () - Paraentheses
- We cannot perform insert, add, remove an item.
```

````python
#Create a empty tuple
empty_tuple = ()

#empty_tuple = tuple()
````

````python
#tuple
fruits = ('Apple','Banana','Mango','lemon')
````

````python
#len() method to get length of tuple
print(len(fruits))
````

````python
#datatype checking
print(type(fruits))
````

```Accessing the elements from a Tuple```
- We can access through indexing in positive and negative as ;ike lists

````python
print(fruits[0]) #fetch the first element
print(fruits[-1]) #Fetch the last element
````

```Slicing tuple```
- we can slice the tuple by specifying the range of indexes by providing the start, end, & step.

````python
print(fruits[0:2]) #apple banana
print(fruits[1:3]) #it will exclude the end element as it is exclusive.
````

````python
print(fruits[-4:]) #all items
print(fruits[-3:-1]) #doesn't include item at the index 3
````

```Changing Tuples to Lists```
- We can change the tuples into list and lists to tuples. When ever we need to modify the tuple we need to convert it to list to modify.

````python
#Syntax
tpl =  ('item1', 'item2', 'item3','item4')
print(type(tpl))

lst = list(tpl)
print(type(lst))
````

````python
#Adding element in the converted list.
lst[0] = "Bad item"
print(lst)
````

```Checking an item in a tuple```
- we can check if an item exist or not in a tuple using in, return boolean.

````python
print('Bad item' in tpl)
````

````python
print("Banana" in fruits)
````

```Joining Tuples```
- We can join two or more tuples using + operator

````python
vegetables = ("Tomato","Potato","Cabbage","Carrot")
fruits_and_vegetables = fruits + vegetables
print(fruits_and_vegetables)
````

```Deleting Tuples```
- It's not possible to remove the items in a tuple but we can delete the tuple itself using DEL keyword.

````python
#Deleting the tuple.
del tpl
````
