$$ Day 9 - Conditionals$$

```
Conditional statments are the control flow where the block code is executed based upon the conditon is the certain condition is True or False, it will move further block of code.

```

```IF condition```
```
- IF statement works like if the given condition is true execute the action asked or else execute other action.

#Syntax:

IF Condition:
   #Do this
ELSE:
   #Do something
```

````python
#Example 1:
a = 3

if a > 0:
    print('A is a Positive Number')
````

```IF-ELSE Condition```
- If the first condition is True it will execute the first block, or else it will execute the second block.

````python
#Example 2:
a = 3

if a > 0:
    print('A is a Positive Number')
else:
    print('A is a Negative Number')
````

```IF-ELIF-ELSE```
- In this conditional statements we can have as many as condition in between if and else block.

````python
a = 0

if a > 0:
    print('A is positive number')
elif a < 0:
    print('A is Negative number')
else:
    print('A is zero')
````

```Nested Conditions```
- These are the conditions inside a condition.

```py
Syntax:
IF condition:
    #Do this
    IF Condition:
        #Do this
```

````python
a = 0

if a > 0:
    if a % 2 == 0:
        print('A is a Positive & Even Number.')
    else:
        print('A is positive')
elif a == 0:
    print('A is zero')
else:
    print('A is Negative number.')
````

- We can avoid Nested conditions by using Logical operators AND/OR/NOT.

```IF Condition and Logical Operators```

```py
#Syntax:

IF Condition1 and condition2:
     #Do this
```

````python
a = 0

if a > 0 and a % 2 == 0:
    print('A is a Positive & Even Number.')
elif a == 0:
    print('A is zero')
else:
    print('A is Negative number.')
````

```IF and Or Logical Operators```

````python
user = 'Rakshith'
access_lvl = 3

if user == 'Admin' or access_lvl >= 5:
    print('Access granted')
else:
    print('Access Denied')
````

````python
#q4. Write a code which gives grade to students according to theirs scores?

score = int(input("Enter your score: "))

if score >= 90:
    print(f'Your Grade is A')
elif score >= 80:
    print('Your Grade is B')
elif score >= 70:
    print("Your Grade is C")
elif score >= 60:
    print('Your Grade is D')
else:
    print('Your Grade is F')
````

````python
#q5. Get the month from user input then check if the season is Autumn, Winter, Spring or Summer. If the user input is: September, October or November, the season is Autumn. December, January or February, the season is Winter. March, April or May, the season is Spring June, July or August, the season is Summer

user_input = input("Enter the month: ").capitalize()

if user_input in ["September", "October", "November"]:
    print("The Season is Autumn")
elif user_input in ["December", "January","February"]:
    print("The Season is Winter.")
elif user_input in ["March", "April","May"]:
    print("The Season is Spring.")
elif user_input in ["June", "July", "August"]:
    print("The Season is Summer.")
else:
    print('Invalid Response.')
````

````python
#q6.If a fruit doesn't exist in the list add the fruit to the list and print the modified list. If the fruit exists print('That fruit already exist in the list')

fruits = ['banana', 'orange', 'mango', 'lemon']

fru_inp = input("Enter the fruits: ")

if fru_inp not in fruits:
    fruits.append(fru_inp)
    print(fruits)
else:
    print('That fruit already exist in the list')
````

````python
#q7.  Check if the person dictionary has skills key, if so print out the middle skill in the skills list?

for p_key,p_val in person.items():
    if p_key == 'skills':
        mid = len(p_val) // 2
        print(f"The middle skill is: {p_val[mid]}")
````

````python
#q8. Check if the person dictionary has skills key, if so check if the person has 'Python' skill and print out the result?

for p_key,p_val in person.items():
    if p_key == 'skills':
        if 'Python' in p_val:
            print(f"The person has the {p_key} named: Python")
````
