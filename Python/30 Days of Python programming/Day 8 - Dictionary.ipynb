$$ Day 8 - Dictionary $$

```
Dictionary: 
- These are collection of unordered, modifiable (mutable) paired (key: value) data type
- Enclosed with {Key:Value}
```

```Creating a Dictionary```

````python
#Empty dictionary
empty_dict = {}

#Dictionary with data values syntax
dct = {'key1': 'value1',
       'key2': 'value2',
       'key3': 'value3',
       'key4': 'value4'}
````

````python
person = {
    'first_name':"Rakshith",
    'last_name':'Kumar',
    'city':'Banglore',
    'skills': ['Python','SQL','ML'],
    'address':{
        'street':'EC',
        'zipcode': '345273'
    }
}

print(person)

#Dictionary length
print(len(person))
````

```Accessing Dictionary Items```
- We can access Dictionary item by referring to its key name.
- If the key name doesn't exist in the dictionary it will raise an error. we can use get() method returns None, which is NoneType if the key doesn't exist.

````python
#Accessing the name key2
print(dct['key2'])
````

````python
#Accessing the city
print(person['city'])
````

````python
#Accessing a key which is not present in the dictionary, by bypassing it through get() to return "None".
print(person.get("Job"))
````

```Adding items to a Dictionary```
- We can add new key and value to a dictionary.

```py
Syntax:
dct = {'key1':'value1', 'key2':'value2', 'key3':'value3', 'key4':'value4'}
dct['key5'] = 'value5'
```

````python
person['Job Title'] = 'Developer'
person['skills'].append("Deep Learning")
print(person)
````

```Modifying items in a Dictionary```
- We can modify item in a dictionary

````python
dct['key2'] = 'Value6'
print(dct)
````

```checking keys in a Dictionary```
- We can use in operator to check if key exists in a dictionary

````python
print('key5' in dct) #False
````

```Removing Key and Value pairs from a dictionary```
- pop(key): remove the item with specified key name
- popitem(): removes the last item
- del dictionary[key]: removes an item with specified key name.

```py
Syntax:
dct.pop('key1') #removes the key1 item.

dct.popitem() #removes the last item from the dictionary.

del dct['key1'] #removes the specified key1 item.
```

```Changing Dictionary to a list of items```
- The items() method changes dictionary to a list of tuples.

````python
#Items() method converts dictionary to list item in tuples.
print(dct.items())
````

```Copy a Dictionary```
- We can copy a dictionary using a copy() method.

````python
dct_copy = dct.copy()
print(dct_copy)
````

```Clearing a Dictionary```
- We can clear the dictionary if we don't want to have it.

````python
dct_copy.clear()
print(dct_copy)
````

```Deleting a Dictionary```

````python
del dct_copy
````

```Getting Dictionary keys & Values as a list```
- The keys() gives us all the keys of a dictionary.
- The values() method gives the values of a dictionary.

````python
#keys
print(dct.keys())
````

````python
#Values
print(dct.values())
````
