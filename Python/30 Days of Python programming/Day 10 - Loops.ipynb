$$ Day 10 - Loops $$

```py

Loop - The process of iterating a block of code until the conditions met all items in a sequenced are processed.

There are two types of loops:

1. For Loop: 
     It is used to iterate over a sequence such as Lists, Tuples, or range.

     #Syntax:

      n = 4

      for i in range(0,n):
        print(i)

2. While Loop:
    It executes a block of code until and unless the condition is true.

    #syntax:

    c = 0

    while c < 3:
        c += 1
        print("line")
```

```WHILE Loop```

```py
It is used to execute a block of statements repeatedly until given condition is satisfied.

#Syntax:
while condition:
   #Code goes here.
```

````python
#Example-1: using While Loop
count = 0

while count < 5:
    count += 1
    print("This example for while loop")
````

- In the above condition, the count is True until and unles the condition met below 5, but right after 5 loop it is greater than 5 where our condition is False so it comes outside while loop.

```While - Else```

- In case if we need run block of code once the condition is no longer True, we can use the Else Block.


```py
#Syntax:
While Condition:
    #Codes goes here
Else:
    #Codes comes here
```

````python
#Example-2
#using While loop with Else

count = 0

while count < 3:
    print("Line count:",count)
    count += 1
else:
    print("While loop ended and came to else block as condition no longer satisfied.")
````

- The above loop condition will be False when count is 3 and loop stops, and execution starts from else statement.

```Break & Continue```
- Break: We use Break when we like to get out of or stop the loop.

```py
#Syntax:

while condition:
    #code comes here
    if another_condition:
        break
```

````python
#Example3: Using while and IF to break

count = 0

while count < 5:
    print("count: ",count)
    count += 1
    if count == 4:
        break
        #Once the if condition is true it will break the loop.
````

```Continue: ```
- With the continue statement we can skip the current iteration, and continue with next step.

```py
#Syntax:
while condition:
    codes goes here
    if another_condition:
        #continue
```

````python
count = 0

while count < 5:
    if count == 3:
        count += 1
        continue
    print("Count",count)
    count += 1
#This will increment and once it equal to 3 it skips the loop and comes out of loop to execute the outside loop stataments.
````

```For```

For Loop is used to iterate over a sequence (That is either a list, tuple, a dictionary, a set, or a string).

```py
#syntax:
for iterator in list:
    code comes here
```

````python
#list
num = [1,2,3,4,5,6]

for i in num:   #i is temporary name to refer to the list items, valid only inside this loop
    print(i)    #This will print one by one
````

```py
#Syntax:

for iterator in string:
    code comes here
```

````python
#string
name = 'Hello'

for i in name:
    print(i)
````

````python
for char in range(len(name)):
    print(name[char])
````

````python
#Tuple
tup = (1,2,3,4)

for i in tup:
    print(i)
````

```Dictionary```
- For loop with dictionary gives you the key of the dictionary.


```py
#syntax:
for iterator in dictionary:
    code goes here
```

````python
person = {
    'first_name': 'Rakshith',
    'last_name': 'Kumar',
    'age': 250,
    'city': 'Banglore',
    'is_married': False,
    'skills': ['SQL', 'ML', 'Deep Learning', 'MongoDB', 'Python'],
    'address': {
        'street': 'Electronic city',
        'zipcode': '456261'
    }
    }

#prints only the keys
for key in person:
    print(key)
````

````python
#to fetch key & value using items() method.
for key,value in person.items():
    print(key,value)
````

````python
#For Loop using Set

st = {1,2,3,4,6}

for i in st:
    print(i)
````

```Break & Continue: Part2```

- Break: When we want to stop our loop before it is completed.

```py
#syntax:
for itreator in sequence:
    code goes here
    if condition:
        break
```

````python
numbers = (1,2,3,4,5,6)

for i in numbers:
    print(i)
    if i == 3:
        break

#In the above code the loop will iterate over the tuple and once it reachs the if condition it will break and exit the loop.
````

- Continue: We us continue when we want to skip some of the steps in the itreation of the loops.

```py
#syntax:

for iterator in sequence:
    code goes here
    if condition:
        continue
```

````python
numbers = (1,2,3,4,5,6)

for num in numbers:
    print(num)
    if num == 3:
        continue
    print("Next number should be",num+1)
print("Outside the loop")
````

```Range() Function```
- The range() function is used to return a list of numbers. The range(start,stop,step) takes three parameters.


```py
for iterator in range(start,end,step):
    code goes here
```

````python
lst = list(range(4))
print(lst) #[0,1,2,3]
````

````python
lst = list(range(1,5,2))
print(lst) #[1,3]
````

```Nested For Loop```
- We can write a loop inside a loop.


```py
#syntax:
for x in y:
    for t in x:
        print(t)
```

````python
person = {
    'first_name': 'Rakshith',
    'last_name': 'Kumar',
    'age': 250,
    'city': 'Banglore',
    'is_married': False,
    'skills': ['SQL', 'ML', 'Deep Learning', 'MongoDB', 'Python'],
    'address': {
        'street': 'Electronic city',
        'zipcode': '456261'
    }
    }

for key in person:
    if key == 'skills':
        for skill in person['skills']:
            print(skill)
````

```For Else```

- if we need to execute some block we use else statament.

```py
#syntax:
for iterator in range(start,stop, step):
    #do something
else:
    print("The loop ended".)
```

````python
#For-Else usage.
for i in range(1,5):
    print(i)
else:
    print("Loop ended.")
````

```Pass```
- In any statement after semicolon, if we don't want to execute the block of code we can use "Pass" to bypass the code.

```py
#syntax:
for iterator in sequence:
    pass
```
