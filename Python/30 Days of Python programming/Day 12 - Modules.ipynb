#### **Import Inbuilt modules**

- We can also import in-built python modules using import keyword. The most used python modules on regular basis are math, datetime, sys, os, random, statistics, collections, json, re.

##### **OS Module**
- Using the python OS module you can automate many operating system tasks.
- This module provides functions like creating, changing current working directory, and removing a directory (folder), fetching its content, changing and identifying the current directory.

```python
#To use the os module we need to import
import os

#creating a directory
os.mkdir('directory name')

#changing the current directory
os.chdir('path/location')

#getting current working directory
os.getcwd()

#removing directory
os.rmdir('directory name')
```

##### **Sys Module**
- The sys module gives python program access to interpreter level information and control -- such as command line arguments, module search points, standard streams, and program exit.

- sys.argv[0] -- always return the name of the script
- sys.argv[1] -- the argument passed from the command line.

```python
#Command line arguments passed to the script
import sys

#to exit sys
sys.exit()

#to know the largest integer variable it takes
sys.maxsize

#To know the path
sys.path

#To know the version of the python you are using
sys.version
```

##### **Statistics Module**
- This module provides the statitsical functions of numerical data. the most popular functions are: Mean, Mode, median, std,...

```python
import statistics

ages = [20, 20, 4, 24, 25, 22, 26, 20, 23, 22, 26]

print(statistics.mean(ages))
print(statistics.mode(ages))
print(statistics.median(ages))
print(statistics.stdev(ages))

```

##### **Math Module**

- This module helps in many mathematical operations.

- We can import all math functions using *

from math import *

```python
import math

print(math.pi)  #To fetch pi value: 3.14
print(math.sqrt(2)) #To calculate square root of 2 is 4.
print(math.pow(2,3)) #power 2 to 3 is 8.
print(math.floor(9.81)) #round to lowest
print(math.ceil(9.81)) #round to highest value.
```

##### **String Module**
- A string module is a usefull module for many purpose

```python
import string

print(string.ascii_letters)
print(string.digits)
print(string.punctuation)
```

##### **Random Module**

- These gives a random numbers between 0 and 0.999..

```python
import random

print(random.randint(0,10))
print(random.random())     #generates from 0.9
print(random.random() * 5) #until 5 but not the 5.
```
