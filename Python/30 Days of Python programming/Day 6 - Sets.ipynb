```
Sets:
- These are the collection of unique unordered and un-indexing elements.
- Enclosed with {}
- Able to perform intersection, union, difference, super set and so on..
```

```py
# To create a empty set

# Empty set syntax
set_item = set()

- we don't use {} as it will create a dictionary
```

````python
#Creation of set
set_item = {1,2,3,4,7,9,11} #set of integer items
fruits = {'Apple','Banana','Grapes','Lemon'} #set of string item
````

````python
#length of the set using len() method
print(len(set_item))
````

##### Accessing items in a set

- We can use the indexing or slicing for sets as they are unordered.
- We can use loops, membership operators.

````python
for item in set_item:
    if item == 9:
        print(item)
        
````

##### Checking an Item
- To check if an item exist we use memebership operator

````python
print('Lemon' in fruits) #True
````

##### Adding items to a set
- We cannot change any items once set is created but we can add additional item using add() method.

````python
fruits.add('Strawberry')
print(fruits)
````

- To add multiple items using update() allows to add multiple items into set but it takes arguments as list.

````python
fruits.update(['Orange','Dragon Fruit','Berries'])
print(fruits)
````

##### Removing Items from a set
- We can remove an item from a set using remove(). if the item doesn't found in the set it will raise a error.
- We have discard() method doesn't raise any errors.

````python
fruits.remove("Dragon Fruit")
print("Dragon Fruit" in fruits)
````

Pop() will randomly removes a item from the set.

````python
fruits.pop()
print(fruits)
````

##### Clearning Items in a set
- If we want to clear or empty the set, we use clear() method.

````python
set_item.clear()
#keep the definition but clears the items in the set.
print(set_item)
````

````python
#delete a set use del keyword.
del set_item
print(set_item) #raise an error as it won't find the set
````

##### Converting List to Set
- We can convert the list to set and set to list, but during conversion from list to set only unique values are reserved in the set.

````python
#Creating a list with duplicate items as list holds duplicate values
lst = ['item1','item2','item3','item4','item4','item5']

lst = set(lst) #Conversion list --> set (No duplicates)

print(lst)
````

```Joining Sets```
- We can join the sets using union() & update() method or | symbol.

Union() method: This method return new set

````python
st1 = {'item1','item2','item3','item4'}
st2 = {'item2','item5','item6'}
st3 = st1.union(st2)  #st3 = st1 | st2
print(st3)
````

Update(): This method inserts a set into a given set.

````python
#All the content for st2 will be added into st1 if duplicate that would be removed.
st1.update(st2) 
print(st1)
````

##### Finding Intersection items
- Intersection return set of items which are in both the sets.

````python
st1.intersection(st2)
print(st1)
````

```Checking Subset and Superset```
- A set can be subset or superset

- Subset: issubset() -
  A subset where a set whose elements are all contained in another set.
- Super: issuperset()
  A set that contains all the elements of another set


````python
st1.issubset(st1)
````

````python
st2.issuperset(st1)
````

```Checking the difference between two sets```
- It return the difference between two sets or using - symbol

````python
st1.difference(st2)
````

```Finding Symmetric Difference between two sets```
- It returns the symmetric difference between two sets. It means that it return a set that contains all items from both sets, except items that are present in both sets.
- In general if the item present in both the set is ignored and rest will be returned.

````python
st1.symmetric_difference(st2)
````

```Joining Sets```
- If the two sets do not have a common item or items we call them disjoint sets.
- We can check if two sets are joint or disjoint() method. If one item exist it will return False

````python
st1.isdisjoint(st2)
````
