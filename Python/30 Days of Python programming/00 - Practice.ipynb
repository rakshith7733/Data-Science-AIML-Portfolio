```python
# Q1. Count Character frequency?

text = 'hello'
freq = {}

for char in text:
    if char in freq:
        freq[char] += 1
    else:
        freq[char] = 1

print(freq)
```

```python
from collections import Counter
freq = Counter(char)
print(freq)
```
