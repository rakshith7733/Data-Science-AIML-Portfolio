````python
# Q1. Count Character frequency?

text = 'hello'
freq = {}

for char in text:
    if char in freq:
        freq[char] += 1
    else:
        freq[char] = 1

print(freq)
````

````python
from collections import Counter
freq = Counter(char)
print(freq)
````

````python
#IF
age = 20
if age == 20:
    print('Execute this block.')
````

````python
#IF-ELIF-ELSE

if age > 20:
    print('This is IF block')
elif age > 18 and age <= 20:
    print("This is ELIF block")
else:
    print('This is ELSE block')
````
