#### ```Functions with Parameters```

- In a function we can pass different data types (number, string, boolean, list, tuple, dictionary or set) as parameters.

```Single Parameter```
- If a function takes a parameter we should pass our calling function with single argument.

```py
#Syntax:

#Defining a function
def function_name(parameter):   #This is our parameter
     #Code

#Calling the function

```

````python
#Function with single parameters.

#Defining a function
def greeting(name):                       #name is our parameter
    word = 'Hello' + ' ' + name
    return word

#calling a function
print(greeting('Rakshith'))                     #'rakshith' is the argument.
````

````python
#Example
def sum_of_num(n):
    total = 0
    for i in range(n+1):
        total += i
    return total

print(f"Sum of numbers: {sum_of_num(100)}")
````

```Function with two parameters```

```py
- The function may or maynot have parameters. if the parameter takes two then we need to call them using the respective arguments in the calling section.

#Defining a function
def function_name(parameter1, parameter2):
    #code

#Calling a function
function_name(argument1,argument2)
```

````python
#Function with two parameters

#defining a function
def name_generator(f_name,l_name):
    full_name = 'Hello' + ' '+ f_name + ' ' + l_name
    return full_name

#Calling a function.
print(name_generator('Rakshith','Kumar'))
````

- In the above code we have given two parameter based on that the same arguments were given, the arguments takes the serial wise the parameter we defined.
- Suppose if we give arguments of first to second and like wise it will take as per parameter wise defined. To handle the issue we can use the argument passing through key, value, In this case the order won't be an issue.

````python
#Function with parameter with key, value arguments.

#Defining a function
def add_two_num(num1,num2):
    sum_of_num = num1 + num2
    return sum_of_num

#Calling a function
print(add_two_num(num2=30,num1=20))    #we can see that argument is postion is different as this can be given through key,value process.
````

````python
#Example for boolean return statement

def check_num(num):
    if num % 2 == 0:
        return True

print(check_num(10))  #True
````

````python
#Example for list return using return statement.

#defining a function
def list_gen(n):
    lst = []
    for i in range(n+1):
        if i % 2 == 0:
            lst.append(i)
    return lst

#Calling a function
print(list_gen(10))
````

```Functions with default parameters```

```py
- Sometimes we pass default values to parameter, when we invoke/call the function if the argument is not given then it will choose the parameter value.

#Syntax:

#Defining a function.

   #code


#Calling a function

```

````python
#Function with default parameter values

#Defining a function
def greet(name='Rakshith'):
    message = 'Hello' + ' '+name
    return message

#Calling a function
print(greet())
print(greet('Plus'))
````

- In the above code we can see that the first calling function has no arguments so it took the default parameter we defined. But in the second calling function we have passed the argument so it replaced with default one.

```Arbitary Number of arguments```

```py
- If we do not know the number of arguments we pass to our function, we can create a function which can take arbitary number of arguments by adding * before the parameter name.

#Syntax:

#Defining a function
def function_name(*args):
    #code

#calling a function
function_name(parmeter1, parameter2,parameter3,....)
```

````python
#Example 1 function with arbitary arguments.

#Defining a function
def sum_of_num(*nums):
    total = 0
    for i in nums:
        total += i
    return total

#Calling a function
print(sum_of_num(3,4,5,6,7,8,9))
````

```Dictionary Unpacking```
- You can call a function when has named arguments using a dictionary with matching key names. you do using **

````python
#Defining a fuction that takes two arguments: 'name' and 'Location'.
def greet(name,location):
    print("Hi there",name, 'how is the weather in',location)

#calling the function with keyword arguments.
greet(name='Rakshith',location='Bengaluru')

#Create a dictionary with keys matching the function's parameter names
my_dict = {'name':'Rakshith','location':'Bengaluru'}

#call the function using dict unpacking
greet(**my_dict)

# The ** operator unpacks the dictionary, passing its key value pairs as keyword arguments to the function.
````

```Arbitary number of Named arguments```

- You can also define a function to accept an arbitary number of named arguments.

- Avoid this unless required as it makes it harder to understand what the function accepts and does.

````python
#Defining a function
def arbitary_named_args(**args):
    print('I received an arbitary number of arguments, totalling',len(args))
    print("They are provided as a dictionary in my function:",type(args))
    print("Let's print them:")

    for k,v in args.items():
        print(" * key:",k,"value: ",v)

#creating a dictionary to unpack
dict = {'name':'Rakshith',
        'location':'Bengaluru',
        'role':'SWE',
        'skills':['python','ML','SQL']
        }

#calling the function
print(arbitary_named_args(**dict))
````

````python
#q8. Declare a function named reverse_list. It takes an array as a parameter and it returns the reverse of the array (use loops)?

def reverse_list(arr):

    start = 0
    end = len(arr) - 1

    while start < end:

        temp = arr[start]
        arr[start] = arr[end]
        arr[end] = temp

        #Increment and decrement
        start += 1
        end -= 1        
    return arr


arr = [45,76,39,56]
print("Reversed_list:",reverse_list(arr))
````

````python
#16. Call your function factorial, it takes a whole number as a parameter and it return a factorial of the number?
def factorial(num):
    fact = 1
    for n in range(1, num+1):
        fact = fact * n
    return fact

print(factorial(12))
````

````python
#20. Create a function called show_args to take an arbitrary number of named arguments and print their names and values?

def show_args(**kwargs):
    for i,j in kwargs.items():
        print(f" {i}: {j}")

dict = {
    'Name':'Alice',
    'Age':22,
    'City':'Belgium',
    'Skills': ['Python','SQL','Machine Learning'],
    'Role': 'Data Scientist'
}


show_args(**dict)
````

````python
#q21. Write a function called is_prime, which checks if a number is prime?

def is_prime(num):

    if num < 2:
        return "Not a Prime"

    for i in range(2, int(num**0.5)+1):
        if num % i == 0:
            print('Not a prime')
            break
    else:
        print("Prime")

is_prime(10)
````

````python
#q22. Write a functions which checks if all items are unique in the list?

def unique_check(lst):
    empty = []
    for item in lst:
        if item in empty:
            print("Not Unique")
            return
        empty.append(item)

    print("Unique")

lst = ["good",'bad',4,6]
unique_check(lst)
````
