#To use the matplotlib features:
- import matplotlib.pyplot as plt

````python
import numpy as np
import matplotlib.pyplot as plt
````

````python
x = np.linspace(0,5,11)
y = x ** 2
````

```Functional Process```

````python
plt.plot(x,y)
plt.xlabel('X Label')
plt.ylabel('Y Label')
plt.title('Title')
````

**Subplot():**
- It allows you to create multiple plot in a single figure in a grid.

#Syntax:
- subplot(nrows, ncoloumns, index)

````python
plt.subplot(1,2,1)
plt.plot(x,y,'r')

plt.subplot(1,2,2)
plt.plot(y,x,'b')
````

```Object Oriented Method```

````python
fig = plt.figure()  #This will create a figure object on this we can design as per need.

axes = fig.add_axes([0.1,0.1,0.8,0.8]) #This will be like dimension of the figure.
axes.plot(x,y)
axes.set_xlabel('X label')
axes.set_ylabel('Y label')
axes.set_title('Title')
plt.show()
````

Steps:
- First create an object of the figure to set an outline.
- add axes to the object as per our dimensions.
- plot the charts you need.

````python
#Creating two plots using Object Oriented method with axes.

#add_axes([left,bottom,width,height]) -- It will control the plot axes.

fig = plt.figure()

axes1 = fig.add_axes([0.1,0.1,0.8,0.8])
axes2 = fig.add_axes([0.2,0.5,0.4,0.3])

axes1.plot(x,y)
axes1.set_title("Larger Plot")

axes2.plot(y,x)
axes2.set_title("Smaller Plot")
````

```Creating a Subplots```
- Figure and grid in one call, It helps in creating multi panel charts.

#Syntax:
> fig, axes = plt.subplots(nrows=1, ncols=2)

> plt.tight_layout() -- adjust or align the charts equally without overlapping.



**Steps:**
- defining the axes and figure.
- using iterate on axes or using index on axes we plot the charts.
- Further design stuff.

````python
axes
#axes are the list of object array. We can iterate over the axes.
````

````python
fig,axes = plt.subplots(1,2)

for current_ax in axes:
    current_ax.plot(x,y)
````

````python
#We can index over the axes
fig,axes = plt.subplots(1,2)

axes[0].plot(x,y)
axes[0].set_title("Axes - 0")

axes[1].plot(y,x)
axes[1].set_title("Axes - 1")
````

````python
fig,axes = plt.subplots(nrows=1,ncols=2)
#plt.tight_layout()
````

````python
import pandas as pd

df = pd.DataFrame({'A': [1,2,3], 'B': [4,5,6], 'C': [7,8,9]})

fig, axes = plt.subplots(1, 3, figsize=(12, 4))
for ax, col in zip(axes, df.columns):   # axes is 1-D here, zips directly
    ax.plot(df[col])
    ax.set_title(col)

plt.tight_layout()
plt.show()
````

````python
#Plotting same charts in same axes
fig = plt.figure()

ax = fig.add_axes([0,0,1,1])
ax.plot(x,y)
ax.plot(y,x)

ax.legend()
````

- For the above two chart plot on one axes we cannot give a legend only itslef, to identify/reference to legend we need to specify the label in the parameter of the plot itself.


- Legend() - what are items present in the chart.
- Use the location code to place the legend on the plot.

````python
fig = plt.figure()

ax = fig.add_axes([0,0,1,1])
ax.plot(x,x**2,label='X Squared')
ax.plot(x,x**3, label='Y Cubed')

ax.legend(loc=0)
````

```Plot Appearance / Setting Colors```

````python
fig = plt.figure()

ax = fig.add_axes([0,0,1,1])

ax.plot(x,y,color='purple',linewidth=2, alpha=0.5, linestyle='-',marker='*', markersize=15,markerfacecolor='red', markeredgewidth=3)

# Color: we can use basic color {orange,blue} or RGB HEX Code {#FF8C00}.
# linewidth: We can adjust the width of the line of our plot like thin/thick. 
# Alpha: we can use alpha to give transperancy of the line.
# Line style: what kind of line we need like dotted(--),dashed(-.), steps
# Marker: To make the visual points to be highlighted like {'o','*','1'}
# Marker size: It correctly set the marker highlighting.
# Markerfacecolor: To give the marker color.
# Markeredgewidth: To thicken the marker borders.
````

```Line and Marker Styles```

- To change the line width, we can use the linewidth or lw keyword argument. The line style can be selected using the linestyle or ls keyword argument.

````python
fig, ax = plt.subplots(figsize=(12,6))

ax.plot(x, x+1, color='red', linewidth=0.25)
ax.plot(x, x+2, color='red', linewidth=0.50)
ax.plot(x, x+3, color='red', linewidth=1.00)
ax.plot(x, x+4, color='red', linewidth=2.00)

#passing linestyle options '-', '--', '-.', ':'. 'steps'
ax.plot(x, x+5, color='green', lw=3, linestyle='-')
ax.plot(x, x+6, color='green', lw=3, ls='-.')
ax.plot(x, x+7, color='green', lw=3, ls=':')

#Custom dash
line, = ax.plot(x, x+8, color='black', lw=1.50)
line.set_dashes([5,10,15,10]) #format: line length, space length,..

#Possible marker symbols: marker = '+', 'o', '*', 's', ',', '.', '1', '2'
ax.plot(x, x+9, color='blue', lw=3, ls='-', marker='+')
ax.plot(x, x+10, color='blue', lw=3, ls='--', marker='o')
ax.plot(x, x+11, color='blue', lw=3, ls='-', marker='s')
ax.plot(x, x+12, color='blue', lw=3, ls='--', marker='1')

#Marker size and color
ax.plot(x, x+13, color='purple', lw=1, ls='-', marker='o', markersize=2)
ax.plot(x, x+14, color='purple', lw=1, ls='-', marker='o', markersize=4)
ax.plot(x, x+15, color='purple', lw=1, ls='-', marker='o', markersize=8, markerfacecolor='yellow')
ax.plot(x, x+16, color='purple', lw=1, ls='-', marker='o', markersize=8, markerfacecolor='yellow', markeredgewidth=3, markeredgecolor='green')
````

```Control over axis apperance```

**Plot Range**
- We can configure the ranges of the axes using the set_ylim and set_xlim methods in the axis object, or axis('tight') for automatically getting 'tightly fitted' axes ranges.

````python
fig, axes = plt.subplots(1,3, figsize=(12,4))

axes[0].plot(x, x**2, x, x**3)
axes[0].set_title("Default axes ranges")

axes[1].plot(x, x**2, x, x**3)
axes[1].set_title()
````

````python
fig = plt.figure()

ax = fig.add_axes([0,0,1,1])

ax.plot(x,y,color='purple',linewidth=2, linestyle='-')

ax.set_xlim([0,1])
ax.set_ylim([0,2])
````
