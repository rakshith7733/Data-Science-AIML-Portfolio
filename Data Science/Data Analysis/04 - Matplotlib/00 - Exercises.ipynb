````python
import numpy as np
import matplotlib.pyplot as plt
````

````python
#Data
x = np.arange(0,100)
y = x*2
z = x**2
````

#Q1. 
- Create a figure object called fig using plt.figure()
- use add_axes to add an axis to the figure canvas at [0,0,1,1]. call this new axis ax.
- plot (x,y) on that axes and set the labels and titles to match the plot.

````python
fig = plt.figure()

axes_ax = fig.add_axes([0,0,1,1])
axes_ax.plot(x,y)
axes_ax.set_xlabel("X Label")
axes_ax.set_ylabel("Y Label")
axes_ax.set_title("Title")
plt.show()
````

- Create the plot by adding two axes to a figure object at [0,0,1,1] and [0.2,0.5,0.4,0.4]?

- Now use x,y and z arrays to recreate the plot. Notice the xlimits and ylimits on the inserted plot

````python
fig = plt.figure()

ax1 = fig.add_axes([0,0,1,1])
ax2 = fig.add_axes([0.2,0.5,0.4,0.4])

ax1.plot(x,z)
ax1.set_xlabel('X')
ax1.set_ylabel('Z')


ax2.plot(x,y)
ax2.set_title("Zoom")
ax2.set_xlabel('X')
ax2.set_ylabel('Y')
ax2.set_xlim([20,22])
ax2.set_ylim([30,50])

plt.show()
````

#Q4.
- Use plt.subplots(nrows=1, ncols=2) to create the plot

````python
fig, axes = plt.subplots(1,2)
````

- Now plot (x,y) and (x,z) on the axes. Play around with the linewidth and style

````python
fig, axes = plt.subplots(1,2)

axes[0].plot(x,y, color='blue', lw=3, ls='--')

axes[1].plot(x,z, color='red', lw=3)

plt.tight_layout() #To avoid overalapping.
plt.show()
````

- see if you can resize the plot by adding the figsize() argument in plt.subplots() are copying and pasting your previous code.

````python
fig, axes = plt.subplots(1,2, figsize=(9,3))

axes[0].plot(x,y, color='blue', lw=3, ls='--')

axes[1].plot(x,z, color='red', lw=3)

plt.tight_layout() #To avoid overalapping.
plt.show()
````
