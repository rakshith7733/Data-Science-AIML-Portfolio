```Indexing & Selection```

````python
import numpy as np
````

````python
arr = np.arange(0,11)
arr
````

````python
#To get an item from the array using index value.
arr[8]
````

````python
#To get items from the array use slice notation 
arr[2:7] #Exclusive of 7 position
````

````python
#slicing
arr[:6] #until position 6 from starting
````

````python
 #starting from 6 position to end
arr[6:]
````

````python
#Broadcasting
arr[0:3] = 25
arr #it will assign the range with assigned value where it replaces the existing value with new value.
````

````python
#This will fetch everything in the array
arr[:]
````

````python
arr = np.arange(0,11)
slice_of_arr = arr[0:6]
slice_of_arr
````

````python
slice_of_arr[:]  = 23
slice_of_arr

#This is just a view of the array but if will affect the orignial array as shown in below
````

````python
arr
#We did the selection from arr with what we need and we assigned with new value for a view it has not copied to new array, which got effected to original array. To avoid memory issue for a large array.
````

````python
#To avoid above issue we use copy
arr_copy = arr.copy()
arr_copy
````

````python
#If we do any broadcasting it will affect only the new one not the original array.
arr_copy[:] = 1
arr_copy
````

Note: If you slice of array and assign a variable it will just take the view but not copy the array.

```2D Array Computations```

````python
#Creation of 2D array
array_2d = np.array([[5,10,15],[20,25,30],[35,40,45]])
array_2d
````

````python
#To fetch the element 5
array_2d[(0,0)] #using tuple method
array_2d[0][0] #
````

````python
#To get entire 1st row
array_2d[1]
````

````python
#To get 40
array_2d[2][1]

#or use comma stating [row,column]
array_2d[2,1]
````

````python
#To fetch only right side of to rows data
array_2d[:2,1:]
````

##### Conditional Selection

````python
arr = np.arange(1,11)
arr
````

````python
#It will return boolean value True/False based on condition matching.
bool_arr = arr  > 5
bool_arr
````

````python
#It will return the values that are True to our condition.
arr[bool_arr]
````

````python
#Conditional selection.
arr[arr>5]
````
