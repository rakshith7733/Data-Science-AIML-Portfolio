- Numpy is a linear algebra library, it is incredible fast works great on numerical computations.

- Numpy arrays comes in two ways: 1. Vectors
                                  2. Matrices

- Vectors are strictly 1-d arrays and matrices are 2-d arrays. Matices can still have only row or columns

````python
#To import numpy library for our numerical computations.
import numpy as np
````

##### Numpy Arrays

````python
#casting list item to numpy array.

#list
lst = [1,2,3]

#ccasting using np.array
arr = np.array(lst)
arr
````

````python
#Type of the variable we cast.
print(type(arr))
````

````python
#We are creating a nexted list and we convert it to 2D array with np.array
my_mat = [[1,2,3],[4,5,6],[7,8,9]]
print(my_mat)
````

````python
#Converting to array it will result in 2D matrice
arr_mat = np.array(my_mat)
arr_mat
````

```py
np.arange() - It result in generate a value range mentioned from start from end. exclusive of end value.

#Syntax:
- np.arange(start,stop,step)
```

````python
#To create a range of values
var1 = np.arange(1,10)
print(var1)

var2 = np.arange(1,10,2)
print(var2)
````

````python
#To get all zeros
np.zeros(3)
````

````python
#To get all ones we have inbuilt function of ones
np.ones(3)
````

- You might aware that we go decimal values of the above code use dtype=int

````python
np.ones(3,dtype=int) #1D array
````

- Suppose if you need matrice we can give using tuple, where one will be number of rows and other will be no of columns

````python
#we are taking 2x3 matrix with only integer values of one
np.ones((2,3),dtype=int)

# 2 is the row
# 3 is the column
````

````python
#linspace() - to get evenly spaced values from range specified based on step. It will result in 1D array.

np.linspace(0,5,10)
````

````python
#Identical matrix - A matrix havng same dimensions and equal to one another.

np.eye(4)  #2D array
````

````python
#For random values uniformly distributed.
np.random.rand(5) #1D array
````

````python
#For random values in 2D array
np.random.rand(2,3)
````

````python
#For random value from a standard normal or gausian distribution.
np.random.randn(4) #1D
np.random.randn(2,3) #2D
````

````python
#To reshape the array. It will result an error if it not full the complete matrix with our dimensions. Ex if 2x2 means all four should be filled or else it will be an error.

arr = np.arange(25)
arr.reshape(5,5)   #5x5=25 elements need to fit in our matrix.
````

````python
#To fetch maximum value
arr.max()
````

````python
#To fetch the minimum value
arr.min()
````

````python
#to find the index of the max and min value
print(arr)

print(arr.argmax()) #maximum index
print(arr.argmin()) #minimum index
````

````python
#To find shape of the array
arr.shape
````

````python
#To find the datatype of the variable
arr.dtype
````
