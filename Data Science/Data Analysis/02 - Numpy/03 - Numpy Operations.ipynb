$$ Numpy Operations$$

```Numpy Operations```
- Array with array
- Array with scalars
- Universal Array Functions

````python
import numpy as np
````

````python
arr = np.arange(0,11)
arr
````

```Array with Array```

````python
#array + array
#This will not combine the two array as list do it actually add with item + item
arr + arr
````

````python
#Muliplication of array x array.
arr * arr

#If you take a look we can see that it is calculating on orginal data but not on the previous calculated values from above tab.
````

```Scalar Operations```
- To performing mathematical operations between a single value (a scalar) and an array, where the operation is applied element-wise to every component of the array.

````python
arr * 10
````

````python
arr + 100
````

````python
arr - 100
````

````python
arr / arr
#In general it should give division by zero error but in numpy it show runtimeWarning error and provide a "Nan" element as result.
````

````python
#While 1 divided by arr gives an "inf" - infinite value as reult with warning sign.
1 / arr
````

````python
arr ** 2
````

```Universal Array Functions```

````python
#Square root
np.sqrt(arr)
````

````python
#max method
np.max(arr)
#arr.max()
````

````python
#min method
np.min(arr)
````

````python
#Sine values
np.sin(arr)
````

````python
#log function
np.log(arr)
````
