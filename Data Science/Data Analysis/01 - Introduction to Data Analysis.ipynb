### ```Data Analysis```
- The process of deriving an insights from data to drive a business decision.

- In day to day life we need to fetch the data from different domains and the data can be of different type like structured, semi-structured, or unstructured. we need to get all the data together in one. As the data will be messy we have to make it ready to for further downstream like ML model, data visualization or any other.

- So here comes the pandas into picture where it is an instensive library of python where it handles the data to shape for further analysis.

```py
Library - A bundle of code written by the developer and shared across the community to solve all kind of common problems.
```

```py
Whenever you get a dataset, ask:


1. Descriptive

    What happened?

    Examples:

    How much did we sell?
    How many customers?
    How many orders?
    Which category sells most?

2. Trend

    How is it changing?

    Sales by month?
    Customers increasing/decreasing?
    Which product is growing fastest?

3. Comparison

    What is better/worse?

    Region A vs Region B?
    Product A vs Product B?
    New vs returning customers?

4. Segmentation

    Who/what behaves differently?

    High-value customers?
    Low-value customers?
    Premium vs regular customers?
    Age groups?

5. Relationship

    What might influence what?

    Discount vs sales?
    Price vs quantity?
    Rating vs sales?
    Delivery time vs customer satisfaction?

6. Anomaly

    What's unusual?

    Sudden sales spike?
    Extremely expensive order?
    Customer with unusually high purchases?
    Product with zero sales?

7. Action

    So what should the business do?

    This is the most important.

    Example:
    "Customers receiving discounts purchase 18% more."

    Don't stop.

    Ask:
    Should we increase discounts?
    Maybe not.

    You need to investigate:

    Is revenue higher?
    Is profit higher?
    Are discounts causing losses?

    That's Analytical thinking.
```
