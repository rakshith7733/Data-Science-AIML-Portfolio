### ```Pandas Exercise SF Salaries```

- The Sanfransico Salaries dataset is pulledd from Kaggle. 
- The link to download: https://www.kaggle.com/datasets/kaggle/sf-salaries

````python
import pandas as pd
````

````python
# Q1. Read salaries data into a dataframe.
sal = pd.read_csv('Salaries.csv')
````

````python
# Q2. Display the dataframe
sal.head()
````

````python
# Q3. Use the info() method to find out how many entries are there.
sal.info()
````

````python
# Q4. We are converting important string column to numeric due to potential analysis.
sal['BasePay'] = pd.to_numeric(sal['BasePay'], errors='coerce')
sal['OvertimePay'] = pd.to_numeric(sal['OvertimePay'], errors='coerce')
sal['OtherPay'] = pd.to_numeric(sal['OtherPay'], errors='coerce')
sal['Benefits'] = pd.to_numeric(sal['Benefits'], errors='coerce')
````

````python
#Q5. What is the average Base Pay?
sal['BasePay'].mean()
````

````python
#Q6. What is the highest amount of OvertimePay in the dataset?
sal['OvertimePay'].max()
````

Q7. What is the job title of JOSEPH DRISCOLL? Note: Use all caps, otherwise you may get answer that doesn't match as there is lower case Joshep driscoll.

````python
sal[sal['EmployeeName'] == 'JOSEPH DRISCOLL']['JobTitle']
````

````python
# Q8. How much does JOSEPH DRISCOLL make include benefits.

sal[sal['EmployeeName'] == 'JOSEPH DRISCOLL']['TotalPayBenefits']
````

````python
# Q9. What is the name of highest paid person (including benefits)?
sal[sal['TotalPayBenefits'] == sal['TotalPayBenefits'].max()]
````

````python
#Advanced method
sal.loc[sal['TotalPayBenefits'].idxmax()]
````

````python
# Q10. What is the name of lowest paid person (including benefits)? Do you notice something strange about how much he or she is paid?

sal[sal['TotalPayBenefits'] == sal['TotalPayBenefits'].min()]
````

````python
#Advanced method
sal.iloc[sal['TotalPayBenefits'].argmin()]
````

````python
# Q11. What was the average(mean) basepay of all employees per year? [2011-2014]
sal.groupby('Year')['BasePay'].mean()
````

````python
# Q12. How many unique job titles are there?
sal['JobTitle'].nunique()
````

````python
# Q13. What are the top 5 most common jobs?

sal['JobTitle'].value_counts().head(5)
````

````python
# Q14. How many job titles were represented by only one person in 2013? E.g. Job titles with only one occurrance in 2013.

sum(sal[sal['Year'] == 2013]['JobTitle'].value_counts() == 1)
````

````python
# How many people have the word chief in their job titles?

sal[sal['JobTitle'].str.contains('.*Chief', case=False)]['Id'].count()
````

````python
# Is there a correlation between length of the job title string and salary?

sal['Title_len'] = sal['JobTitle'].apply(len)
````

````python
sal[['Title_len','TotalPayBenefits']].corr()
````
