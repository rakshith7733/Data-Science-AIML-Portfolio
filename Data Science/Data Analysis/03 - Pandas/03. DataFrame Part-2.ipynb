### ```Missing Data```

````python
import numpy as np
import pandas as pd
````

````python
#Creating a Dictionary with null values.
d = {'A': [1,2,np.nan],
     'B':[5,np.nan,np.nan],
     'C':[1,2,3]
}

#We are converting the dictionary into Dataframe
df = pd.DataFrame(d)

#Display of the dataframe
df
````

```py
**Missing Values:**
- Missing values indicates that the data is not avilable.
- It can be NaN, None, pd.NA (newer version indicator).


#Detect Missing Values
df.isnull() / isna() - returns True where values are missing.

#Remove missing values
1. drop rows: It will remove the rows having null values
  df.dropna()

2. Drop Columns: Remove columns that contain missing values.
    df.dropna(axis=1)

3. Drop Rows only when all values are missing
    df.dropna(how='all)
```

````python
#dropping null values
df.dropna() # This method helps in removing all null values in the rows. It's a view not implise on original until you use inplace=True.


#By default the axis=0
````

````python
#Dropping null values in column wise search using axis=1
df.dropna(axis=1)
````

````python
df.dropna(how='all') #This won't work as all the rows need to be null to make it work.
````

````python
#Using Threshold parameter to handle missing values. The rows having threshold of 2 are skipped.
df.dropna(thresh=2)
````

```py
**Fill Missing Values**

#fill with constant value
df.fillna(0)


#fill numeric columns with mean
df['age'] = df['age'].fillna(df['age'].mean())

#fill with median
df['age'] = df['age'].fillna(df['age'].median())

#fill with mode
#mostly used with categorical columns.
df['name'] = df['name'].fillna(df['name'].mode()[0])
```

````python
df.fillna(value='fill_value')       #This will fill the value where it finds the missing values.
df['A'].fillna(value=df['A'].mean())      #This will fill with mean of the column.
df['B'].fillna(df['B'].median())    #This will fill the column with median.
````

Q1. How do you handle missing values?

- First we need to identify the missing values using isnull()
- Analyse the columns and distribution of the data across the columns.
- For numerical use mean, median imputation, and for categorical use mode imputation.
- drop the rows or columns which are having missing values more than 60,80%.
- validate the impact as it should not be bias.

#### ```Groupby```

```py
- GroupBy allows you to group together rows based off of a column or many and perform an aggregate function on them. 
- By itself it will return a groupby object

- Groupby only creates a group and on those group we need to perform opertions like mean(),sum(), max()
- Use a variable when you think that you wuld perform multiple calculations. most used for reusability and readability.

#Single column aggrgation
df.groupby('company')['sales'].sum()

#Multiple Aggregations. will pass aggregation through list items.
df.groupby('Department')['Salary'].agg(['sum','max','mean'])


```

````python
#Creating a dictionary data
data = {'company':['GOOG','GOOG','MSFT','MSFT','FB','FB'],
        'person':['Sam','Charlie','Amy','Vanessa','Carl','Sarah'],
        'sales':[200,120,340,124,243,350]
}

#Converting Dictionary to dataframe
df1 = pd.DataFrame(data)

#Display dataframe
df1
````

````python
#Ex1: Grouping them based on company.
df1.groupby('company') #This will create a groupby object. We need to use aggregate function on the columns.
````

````python
df1.groupby('company')['sales'].sum()
````

````python
#Multiple aggregation example

# mean & Standard deviation of sales based on company
df1.groupby('company')['sales'].agg(['mean','std'])
````

````python
#based on above condition if we need only Google
df1.groupby('company')['sales'].agg(['mean','std']).loc['GOOG']
````

````python
#Count
df1.groupby('company').count()
````

````python
#max on categorical column returns alphabetical order fetching on groupby items.
df1.groupby('company').max()

#If we close look into result we can see that sales are irrespective we only defined based on company.
````

````python
#Statistical information on the dataframe using describe method.
df1.groupby('company').describe()
````

#### ```Joining, Merging and Concatenation```

##### ```Concatenation:```
- Concatenation glues together dataframes, both the dimensions shoild match along the axis you are concatenating on.
- use pd.concat and pass list of dataframes.

Syntax:
pd.concat([df1,df2,df3])

````python
pd.concat([df1,df2,df3])
````

````python
#if we use axis parameter and look with column wise then the empty cells will be null values.
pd.concat([df1,df2,df3],axis=1)
````

````python
#Example Dataframe
left = pd.DataFrame({
    'key':['K0','K1','K2','K3'],
    'A':['A0','A1','A2','A3'],
    'B':['B0','B1','B2','B3']})

right = pd.DataFrame({
    'key':['K0','K1','K2','K3'],
    'C':['C0','C1','C2','C3'],
    'D':['D0','D1','D2','D3']})
````

````python
left
````

````python
right
````

##### ```Merging```
- The merging function allows you to merge Dataframes together using a similar logic as merging SQL tables together.
- Key acts as a primary key for both tables to join.

````python
pd.merge(left,right, how='inner',on='key')
````

````python
left = pd.DataFrame({
    'key1':['K0','K0','K1','K2'],
    'key2':['K0','K1','K0','K1'],
    'A':['A0','A1','A2','A3'],
    'B':['B0','B1','B2','B3']})

right = pd.DataFrame({
    'key1':['K0','K1','K1','K2'],
    'key2':['K0','K0','K0','K0'],
    'C':['C0','C1','C2','C3'],
    'D':['D0','D1','D2','D3']})
````

````python
# Merging two columns if you have multiple primary keys.
pd.merge(left,right, on=['key1','key2'])
````

````python
# OUter merge
pd.merge(left,right, how='outer',on=['key1','key2'])
````

````python
#right merging
pd.merge(left,right,how='right',on=['key1','key2'])
````

##### ```Joining```
- Joining is a convinent method for combining the columns of two different indexed dataframes into a single resulted dataframe.

````python
left = pd.DataFrame({
    'A':['A0','A1','A2'],
    'B':['B0','B1','B2']},
    index=['K0','K1','K2'])

right = pd.DataFrame({
    'C':['C0','C1','C2'],
    'D':['D0','D1','D2']},
    index=['K0','K2','K3'])
````

````python
left.join(right)
````

````python
#Outer Join
left.join(right, how='outer')
````

````python
#right join
left.join(right, how='right')
````
