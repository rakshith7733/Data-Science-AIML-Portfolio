#### ```Working with SQLALCHEMY```

````python
# It allow us to create a simple SQL environment in memory.
from sqlalchemy import create_engine
````

````python
#We are creating a environment in memory to do our processing.
engine = create_engine('sqlite:///:memory:')
````

````python
#We are writing to sql from our dataframe and give connection to memory.
df.to_sql('my_table', engine, index=False)
````

````python
sqldf = pd.read_sql('my_table',con=engine)
sqldf
````
