$$ DataFrames $$

DataFrame: These are two-dimensional and labeled datastructures of different datatypes. It provides efficient data manipulation, data cleaning, aggregations, and Analysis.

- It is similar to excel spreadsheet
- It integrates with other libraries like Numpy, Matplotlib for core building block for Visualization, Aggregations and working with data.
- It handles data from execl (read_excel), CSV (read_csv) format and SQL Databases.

##### ```Selection```

```py
.loc[] is a label-based indexer in pandas. It is used to select the rows/columns by their labels (names) not by position.

Syntax:
- df.loc[row_label,col_label]


#Selecting single row
df.loc['Column_name']

#Selecting multiple rows using list passing.
df.loc[['Col1','Col2']]

#To fetch specific row and column
df.loc['row','col']

#Selecting multiple columns
df.loc[:,['Column_name']]

#Selecting with conditions
df.loc[df['age'] > 25]

#Filtering and selecting specific column
df.loc[df['age] > 25,['Column_name']]

#Update the value
df.loc['A','age'] = 40


#Slicing with labels
df.loc['A':'B']
~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~
```

```py
.iloc[] stands for Integer location and is used to select rows and columns based on their position numbers (Index Positions)

Syntax:
df.iloc[row_position, column_position]


#1. Selecting a single row
df.iloc[1]

#2. Selecting Multiple rows passing with list items.
df.iloc[[0,2]]

#3. Selecting single value
df.iloc[1,1]

#4. Selecting specific rows and columns
df.iloc[[0,2],[0,1]]

#5. Slicing Rows
df.iloc[0:2]

#6. Slicing rows and columns
df.iloc[[0:2],[0:2]]

#7. All rows and specific columns
df.iloc[:,[0:2]]

#8. Last row
df.iloc[-1]

```

```
loc = Location by label (row/column names)
iloc = integer location (row/column position)
```

````python
#Rows label based
df.loc['C']
````

````python
#Same above using iloc. We can see that C is in 2 position.
df.iloc[2]
````

````python
#Subset
df.loc['B','Y']
````

````python
#subset of A,B and W,Y
df.loc[['A','B'],['W','Y']]
````

#### ```Conditional Selection```

````python
#If we use conditional on dataframe it will return boolean type dataframe, where a value meet the condition be True and not met value will be False.
df > 0
````

````python
#we are assigning the bool value to new dataframe to get the actual condition.

booldf = df > 0
booldf
````

````python
#If we take our original dataframe and give this conditional dataframe it will return the result with values and not met condition values with NAN.
df[booldf]
````

````python
#We are using a column of the dataframe to validate with our condition. It will return a series of boolean expressions for the values of data.
df['W'] > 0
````

````python
#When look into a dataframe with specific condition on the column it only return which met with condition rest are skipped.
df[df['W'] > 0]
````

````python
#Fetch rows whose value is less than zero.
df[df['Z'] < 0]
````

````python
#Fetch the data where column Z is greater than 0 and only need X,Y columns.
df[df['W']>0][['X','Y']]
````

- To check with multiple conditions in Pandas. 
- Use parenthesis for each condition and if you are using "AND" then replace with "&". 
- As "AND" take single condition so it will be ambigious in series conditions.

    - AND for & 
    - OR for |

````python
#Multiply condition.
df[(df['W'] > 0) & (df['X'] > 0)]
````

##### ```Multi-Level indexing (Hierachical Indexing)```

```py
Multi-Level Indexing allows a Dataframe to have more than one index level on rows or columns.
```

````python
#Index levels
outside = ['G1','G1','G1','G2','G2','G2']
inside = [1,2,3,1,2,3]
hier_index = list(zip(outside,inside))
hier_index = pd.MultiIndex.from_tuples(hier_index)
````

````python
hier_index
````

````python
df1 = pd.DataFrame(randn(6,2),hier_index,['A','B'])
df1
````

````python
#If we need to access any elements in the dataframe. we can call outside index and go step by step inside the dataframe.
df1.loc['G1'].loc[1]
````

````python
#By default there are no names for the index it will be none
df1.index.names
````

````python
#We can assign new index names to the dataframe by passing through the list items
df1.index.names = ['Groups','Numbers']
````

````python
df1
````

````python
#Example in access an element: 0.072960 from a multi level index dataframe.

#df1                            #Dataframe level
#df1.loc['G2']                  #Outer index
#df1.loc['G2'].loc[2]           #inside index
df1.loc['G2'].loc[2].loc['B']   #Item level

# Dataframe level >> outerindex >> inside index >> item level.
````

- Use cross section (xs) for selecting data from specific level.

Syntax:
df.xs(index value, level=inside index)

````python
#Select the data of Numbers '1'.
df1.xs(1,level='Numbers')
````
