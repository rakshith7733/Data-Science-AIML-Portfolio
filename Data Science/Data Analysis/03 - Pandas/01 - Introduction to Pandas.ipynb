### ```Series```

- A Pandas series is a one dimenisional labeled array that can be store data of any type. It consists of values and an associated index.
- Series is used when working with a single column of data such as Sales, Salaries, Stock prices, sensor readings, or Dataframe columns.
- It provides fast data manipulation, indexing, filtering, and statistical operations making it a fundamental building block for data analysis and ML workflows.

- Earlier data was stored in lists and it was difficult to analysis due to No labels, No statistical operations. when we need to check combined columns.
- Here Series comes into Pictures where we can combine two columns to perform analysis. These mostly work with only one columns

```
Series : One feature / Column
Dataframe : Multiple Features / columns

Series --> Distribution analysis
DataFrame --> Relationship analysis
```

````python
labels = ['A','B','C']
my_data = [10,20,30]
arr = np.array(my_data)
d = {'A':10,'B':20,'C':30}
````

````python
#Series on our list data
pd.Series(data=my_data)
````

````python
#we can use the index as per requirement with index parameters
pd.Series(data=my_data,index=labels)
````

````python
#If we pass array it will convert into series
pd.Series(arr,labels)
````

````python
#If we pass Dictionary it will automatically use the Key as the Index values.
pd.Series(d)
````

````python
#Creating a series
ser1 = pd.Series([1,2,3,4],['USA','Germany','USSR','Japan'])
ser1
````

````python
ser2 = pd.Series([1,2,5,4],['USA','Germany','Italy','Japan'])
ser2
````

````python
#To access the series element. We can access it only by index.
ser1['USSR']
````

````python
#When we try to concate two series it will check the matching elements and sum it up and if the element is not present it will result wil "NAN" and the sum up value will be converted to float datatype.

ser1 + ser2
````
