````python
#Importing necessary libraries
import pandas as pd
import numpy as np
````

````python
#We are converting our categorical values into numeric patterns. We can do labelEncoder.
dum_df = pd.get_dummies(df, drop_first=True)
````

````python
#We are dividing the data into Feature variables and target variables.
X = dum_df.drop("charges", axis=1)
y = dum_df['charges']
````

#### ```Linear Regression with Polynomial Features```

````python
#Importing necessary libraries for model selection and feature extraction.
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
````

````python
#We are creating an object for our polynomial features with Degree = 2
poly = PolynomialFeatures(degree=2)
````

````python
X_poly = poly.fit_transform(X)
````

````python
#We are importing train test split library
from sklearn.model_selection import train_test_split
````

````python
X_train, X_test, y_train, y_test = train_test_split(X_poly,y, test_size=0.3, random_state=2026)
````

````python
#Defining a regressor object for our model
regressor = LinearRegression()
````

````python
#Training our data to our LR model.
regressor.fit(X_train,y_train)
````

````python
#We are predicting on our X_test data
y_pred = regressor.predict(X_test)
````

````python
#We will look into the M = Co-efficient our alogorithm.
print(regressor.coef_)
````

````python
#We will look for C = intercept of our algorithm.
print(regressor.intercept_)
````

````python
#Importing our metrics library.
from sklearn.metrics import r2_score
from sklearn.metrics import mean_squared_error
from sklearn.metrics import mean_absolute_error
````

````python
#Importing mean_squared_error, mean_absolute_error, r2_Score from sklearn.metrics
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score

#printing Mean Squared Error (MSE)
print("MSE -->",np.sqrt(mean_squared_error(y_test, y_pred)))

#Printing Mean Absolute Error (MAE)
print("MAE -->",mean_absolute_error(y_test, y_pred))

#Printing R2 Score
print("R2 Score -->",r2_score(y_test, y_pred))
````

#### ```Using K-Fold Cross Validation```

````python
from sklearn.model_selection import KFold
from sklearn.model_selection import cross_val_score
````

````python
#We are defining our KFold to be split in 5 folds, and shuffle data.
kfold = KFold(n_splits=5, random_state=2026, shuffle=True)
````

````python
#Apply the regressor and calculating result using cross_val_score() method and setting CV=kfold, scoring = 'r2'
results = cross_val_score(regressor, X, y, cv=kfold, scoring='r2')
````

````python
#Assigning a variable R2 to the result.
R2 = results
````

````python
#Printing R2score for 5 folds
print(R2)
````

````python
#Printing mean of R2 Score
print("R-Squared: %.2f"%(R2.mean()))
````

- This is not a quite a good R2 score we have got

#### ```GridSearch with Respect to LASSO Regression on Polynomial Features```

````python
from sklearn.linear_model import Lasso
from sklearn.model_selection import GridSearchCV
````

````python
#We are creating a parameter variable for our alpha in dict format
parameters = dict(alpha=[0,0.001, 0.1, 0.5, 0.8, 1, 1.5, 2])
````

````python
#We are creating our object Lasso() algo.
clf = Lasso()
````

- Here we will use Polynomial features with Lasso with respect to KFold GridSearch

````python
#We are defining our KFold parameters to split data.
kfold = KFold(n_splits= 5, shuffle=True, random_state=2022)
````

````python
#Creating an object for GridSearchCV with Lasso, parameters that defined above, with folds in cv. Scoring based on R2.
cv = GridSearchCV(estimator=clf, param_grid=parameters, scoring='r2', cv=kfold)
````

````python
#Split with polynomial feature training data.
cv.fit(X_poly, y)
````

````python
#we are storing into Dataframe.
df_cv = pd.DataFrame(cv.cv_results_)
````

````python
df_cv
````

````python
print("Lasso with Poly Features:")
print("Best_params: ", cv.best_params_)
print("Best Score: ", cv.best_score_)
````

````python
print("Lasso without Poly Features:")
print("Best_params: ", cv.best_params_)
print("Best Score: ", cv.best_score_)
````

- The above without poly features is the score where I changed the X_poly to X data and calculating the score.
