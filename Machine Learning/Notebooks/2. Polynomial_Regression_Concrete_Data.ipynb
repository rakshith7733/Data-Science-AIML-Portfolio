````python
X = df.iloc[:,:8]
y = df.iloc[:,8]
````

````python
y
````

````python
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
````

````python
#We set our degree for 2 for the polynomial variable
poly = PolynomialFeatures(degree=2)
````

````python
#We need to fit our polynomial variable to be fit on X training data.
X_poly = poly.fit_transform(X)
````

````python
X_poly
````

````python
original_features = X.columns
original_features
````

````python
features_names = poly.get_feature_names_out(input_features=original_features)
````

````python
#Display the results
print("Original features: ", original_features)
print("Generated Features names: ", features_names)
print("Transformed data shape: ", X_poly.shape)
````

````python
#View as a pandas dataframe
df_poly = pd.DataFrame(X_poly, columns=features_names)
````

````python
df_poly.head()
````

- now will split the data into train and test split

````python
from sklearn.model_selection import train_test_split
````

````python
X_train, X_test, y_train, y_test = train_test_split(X_poly,y, test_size=0.2, random_state=42)
````

````python
regressor = LinearRegression()
regressor.fit(X_train, y_train)
````

````python
y_pred = regressor.predict(X_test)
````

````python
y_pred
````

````python
#Coefficients
print("Co-efficients: ",regressor.coef_)

#Intercept
print("Intercept: ",regressor.intercept_)
````

````python
#Importing mean_squared_error, mean_absolute_error, r2_Score from sklearn.metrics
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score

#printing Mean Squared Error (MSE)
print("MSE -->",np.sqrt(mean_squared_error(y_test, y_pred)))

#Printing Mean Absolute Error (MAE)
print("MAE -->",mean_absolute_error(y_test, y_pred))

#Printing R2 Score
print("R2 Score -->",r2_score(y_test, y_pred))
````
