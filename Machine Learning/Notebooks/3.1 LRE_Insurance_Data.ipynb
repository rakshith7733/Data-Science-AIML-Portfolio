### ```LRE (Medical Insurance Data)```

````python
import pandas as pd
import numpy as np
````

````python
df = pd.read_csv(r'C:\Users\raksh\OneDrive\Documents\Data-Science-AIML-Portfolio\Machine Learning\Datasets\insurance.csv')
````

````python
#Quick Inspection
df.head()
````

````python
#To check for null values
df.isnull().sum()
````

````python
#To check for the datatypes, memory usage, and columns
df.info()
````

````python
dum_df = pd.get_dummies(df, drop_first=True)
dum_df.head()
````

````python
#To understand the columns after dummies where categorical converted to boolean
dum_df.info()
````

````python
#We get the list of columns of the dataframes
dum_df.columns
````

````python
#Divide the data into Independent and target variable
X = dum_df.drop('charges', axis=1)
y = dum_df['charges']
````

````python
#Importing train test split function from sklearn.
from sklearn.model_selection import train_test_split
````

````python
#splitting data into train set and test set.
X_train, X_test, y_train, y_test = train_test_split(X,y, test_size=0.20, random_state=2026)
````

#### ```Linear Regression```

````python
#Importing our Linear Regression Algorithm.
from sklearn.linear_model import LinearRegression
````

````python
#Creating an object for our Linear model.
regressor = LinearRegression()
````

````python
#We are using our object to train our algorithm on our splitted training data.
regressor.fit(X_train, y_train)
````

````python
#Understanding the coefficients
print("Co-efficients: ", regressor.coef_)
````

````python
#Checking for Intercept of our data
print("Intercept: ", regressor.intercept_)
````

````python
#Now we are predicting on test data
y_pred = regressor.predict(X_test)
````

````python
#Importing mean_squared_error, mean_absolute_error, r2_Score from sklearn.metrics
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score

#printing Mean Squared Error (MSE)
print("MSE -->",np.sqrt(mean_squared_error(y_test, y_pred)))

#Printing Mean Absolute Error (MAE)
print("MAE -->",mean_absolute_error(y_test, y_pred))

#Printing R2 Score
print("R2 Score -->",r2_score(y_test, y_pred))
````

#### ```Polynomial Regression```

````python
#importing our polynomial model
from sklearn.preprocessing import PolynomialFeatures
````

````python
#We are checking how degree = 2 works with original and generated features.
original_features = X.columns
original_features
````

````python
#Generated features for degree = 2
generated_features = poly.get_feature_names_out(input_features=original_features)
generated_features
````

````python
X_poly.shape
````

````python
#We are creating an Linear model object for our Polynomial algorithm.
regressor = LinearRegression()
````

````python
#Training the data
regressor.fit(X_train, y_train)
````

````python
#Checking for Intercept of our data
print("Intercept: ", regressor.intercept_)
````

````python
#Now we are predicting on test data
y_pred = regressor.predict(X_test)
````

````python
#Importing mean_squared_error, mean_absolute_error, r2_Score from sklearn.metrics
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score

#printing Mean Squared Error (MSE)
print("MSE -->",np.sqrt(mean_squared_error(y_test, y_pred)))

#Printing Mean Absolute Error (MAE)
print("MAE -->",mean_absolute_error(y_test, y_pred))

#Printing R2 Score
print("R2 Score -->",r2_score(y_test, y_pred))
````
