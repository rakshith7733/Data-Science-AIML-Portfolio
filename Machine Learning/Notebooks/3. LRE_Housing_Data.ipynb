## ```LRE Regression```

### ```Example (House Price Prediction)```

- Assume we want to predict House Price using these features:

    1. Size of house (sqft)
    2. Number of rooms
    3. Distance from city center.
    4. Wall color (red/blue) --> Not important
    5. Owners' Lucky Number --> Not important

- Some features are usefull, while some are useless in nature.
-------------------------------------------------------------------

#### **1. Lasso Regression (L1)**

**What is Lasso does?**
- Lasso tries to remove useless features completely.

**Example Outcome:**
- After Training the model:

```
Feature         Coefficient
Size                45
Rooms               30
Distance           -20
Wall Color           0
Lucky Number         0
```

**Explanation in Simple Words:**
- Lasso decided that **wall color** and **lucky number** do not matter.
- Their weights became **zero**.
- The model ignores them completely.

**When this is usefull**
- When you have many features.
- When you want the model to automatically choose important features ones.


- Sometimes it leads to overfitting due to some features are removed by coefficients to zero that leads to some coorelation between them.

-------------------------------------------------------------------
#### **2. Ridge regression (L2)**

**What Ridge does?**
- Ridge tries to **reduces the effect of features**, but it does not remove any.

**Example Outcomes:**

```
Features        Coefficients
Size               30
Rooms              20
Distance          -15
Wall color          2
Lucky Nymber        1
```

**Explanation in simple words**
- All features are still used.
- Even useless features get small weights.
- The model is more stable and less overfitted.

**When this is useful**
- When all features have some importance.
- When features are related to each other.
-------------------------------------------------------------------

#### **3. Elastic Net (L1 + L2)**

**What does Elastic net does?**
- Elastic net removes useless features and shrinks the important ones.

**Example Outcome:**

```
features            Coefficients
Size                   35
Rooms                  25
Distance              -18
Wall Color              0
Lucky Number            0
```

**Explanation in Simple words**
- Useless features are removed.
- Important features are controlled.
- Model is balanced and stable.
-------------------------------------------------------------------

#### **4. Key Difference using the SAME example**

```
Model                       What Happenend
Lasso        --------   Removed wall color and lucky number
Ridge        --------   Kept all features but reduced impact
Elastic Net  --------   Removes useless ones and shrinks others.
```
-------------------------------------------------------------------

#### **5. Short Interview-friendly Explantion**
- **Lasso** removes unnecessary features by setting coefficients to zero.
- **Ridge** keeps all features but reduces their influence.
- **Elastic Net** combines both by removing useless features and stabilizing important ones.

````python
import pandas as pd
import numpy as np
````

````python
#Reading the data from CSV format.
df = pd.read_csv(r'C:\Users\raksh\OneDrive\Documents\Data-Science-AIML-Portfolio\Machine Learning\Datasets\Housing.csv')
````

````python
df.head()
````

````python
#To check for null values
df.isnull().sum()
````

````python
#To check for the datatypes, memory usage, and columns
df.info()
````

````python
#We are applying get dummies on our dataframe so that our categorical values were converted to numerical value by dropping the first instances(drop_first=True)
dum_df = pd.get_dummies(df, drop_first=True)
````

````python
dum_df.head()
````

````python
#To understand the columns after dummies where categorical converted to boolean
dum_df.info()
````

````python
#We get the list of columns of the dataframes
dum_df.columns
````

````python
#We are creating two variables to store our Target variable as 'y' and input variables as 'X'
y = dum_df.iloc[:,0]
X = dum_df.iloc[:,1:12]
````

````python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X,y, test_size=0.20, random_state=42)
````

#### ```Lasso regression (L1)```

````python
from sklearn.linear_model import Lasso
````

````python
clf = Lasso(alpha=3, selection='random') #We are taking alpha as 3 as we can take from (0 - inf)
````

- Alpha value is controlling the intercept values. It make values closer to our alpha.

````python
clf.fit(X_train, y_train)
````

````python
print(clf.coef_)
````

````python
print(clf.intercept_)
````

````python
y_pred = clf.predict(X_test)
````

````python
#Importing mean_squared_error, mean_absolute_error, r2_Score from sklearn.metrics
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score

#printing Mean Squared Error (MSE)
print("MSE -->",np.sqrt(mean_squared_error(y_test, y_pred)))

#Printing Mean Absolute Error (MAE)
print("MAE -->",mean_absolute_error(y_test, y_pred))

#Printing R2 Score
print("R2 Score -->",r2_score(y_test, y_pred))
````

- We can conclude that Lasso Regression is not a good fit for our model.

#### ```Ridge Regression (L2)```

````python
from sklearn.linear_model import Ridge
````

````python
rdg = Ridge(alpha=3)
````

````python
rdg.fit(X_train, y_train)
````

````python
print(rdg.coef_)
````

````python
print(rdg.intercept_)
````

````python
y_pred = rdg.predict(X_test)
````

````python
#Importing mean_squared_error, mean_absolute_error, r2_Score from sklearn.metrics
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score

#printing Mean Squared Error (MSE)
print("MSE -->",np.sqrt(mean_squared_error(y_test, y_pred)))

#Printing Mean Absolute Error (MAE)
print("MAE -->",mean_absolute_error(y_test, y_pred))

#Printing R2 Score
print("R2 Score -->",r2_score(y_test, y_pred))
````

- We can also consider that Ridge is alos not a good fit for our model.

#### ```Elastic Net Regression (L1+L2)```

````python
from sklearn.linear_model import ElasticNet
````

````python
enr = ElasticNet(alpha=3, l1_ratio=0.6)
````

````python
enr.fit(X_train, y_train)
````

````python
print(enr.coef_)
````

````python
print(enr.intercept_)
````

````python
y_pred = enr.predict(X_test)
````

````python
#Importing mean_squared_error, mean_absolute_error, r2_Score from sklearn.metrics
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score

#printing Mean Squared Error (MSE)
print("MSE -->",np.sqrt(mean_squared_error(y_test, y_pred)))

#Printing Mean Absolute Error (MAE)
print("MAE -->",mean_absolute_error(y_test, y_pred))

#Printing R2 Score
print("R2 Score -->",r2_score(y_test, y_pred))
````

- We can consider this model doesn't fit good with housing data.
