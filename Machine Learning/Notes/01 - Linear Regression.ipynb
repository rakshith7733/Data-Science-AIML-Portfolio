$$ Linear~Regression$$

- Linear Regression is a supervised machine learning algorithm where it is used to predict continuous numerical values. It finds the best fit line between the independent variables and target variables by minimizing the errors between actual and predicted values. The co-efficients are adjusted to reduce the loss.

- Equation: y = mx + c

- Where: y = predicted value,
       x = input
       m = slope
       c = intercept


- The Y is directly proportional to X in the axis of linear regression problem. Ex: if a area is increasing then the price of the house will also be increasing.

- If they are many different independent variables those elements will contribute in slope and intercept for the prediction and the equation will be Yp = Mx + bz + C

- Taking House pricing example with different independent variables then Y is price x is Area, Z is No of Rooms.
- We can have as many as independent varaiables but our predicted outcome should be only one (What actually we need to find through the equation).


- Our goal with linear regression is to minimize the vertical distance between all the data points and our line.


Example:
If a line starting from zero in X & Y axis, then the linear equation will be y = mx. Not every line starting from (0,0) will be having 'C' intecept. Intercept is calculated from the data through patterns where it is necessary to have an intercept.

##### **Assumptions in Linear Regression**
```
1. Linearity          -- Relationship between input and output is a straight line.
2. Indepedence         -- Inputs should be independent of each other.
3. Homoscedasticity    -- Same spread of errors (no funnel shape).
4. No Multicollinearity-- Inputs shouldn't be too related to each other.
5. Normality of errors -- Prediction errors should follow a bell curve.
```

##### **Implementation of Linear Regression (Steps)**

1. Collect input and output data.
2. Fit a straight line using y = mx + c
3. use loss function like (Mean Absolute Error)MAE or (Mean Squared Error)MSE to reduce error.
4. Predict new values.

#### ```Polynomial Regression```

##### Introduction:
- Used when the data is not linear -- It curves.
- Still uses regression, but now it fits a curved line instead straight one.

##### Why Ploynomial regression?
- Linear Regression can't handle curved relationships.
- Polynomial regression adds extra powers of inputs like X^2, x^3.

- Equation: Y = a + b1* x + b2*x^2 +b3*x^3 + .... + bn*x^n.


##### How does it handle non-linear data?
- By adding more **"curves"** using powers of x, the model can bend and fit the shape of the data better.

```py
Degree             Meaning in simple words
  1                 Linear (straight line)
  2                 Quadratic (U shaped curve)
  3                 Cubic (One hump or dip)
  4+                Wavy lines - fits more complex data patterns 

<> Higher degree = more flexibility, but can also overfit (too tightly hugging the data).
```

##### **Implementation Steps:**
1. Create new features: x^2, x^3, ..., x^n
2. Fit a Linear Regression Model on these new features.
3. Predict values using the new equation.


##### **How to use and when:**


- Regression -- predict 1 value
- Classification -- predict >2 values
- Forecast -- Forecast 1 value. (comes under Regression as well).
