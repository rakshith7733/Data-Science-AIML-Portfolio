### ```Cross Validation```

#### **1. Why do we need Cross Validation?**

When we train a machine learning regression model, we usually do:
- Train model on training data.
- Test on Test Data.
- Check Accuracy (R^2, RMSE, MAE, MSE,..)

But this approach has a problem.

##### **Problem**
If the dataset is small, the model may **overfit** the training data.

**Example:**

|    Data          |  Prediction  |
|------------------|------------
|Training Accuracy |      95%   
|Test Accuracy     |      60%   

- This means the model **memorized the training data**, not learned patterns.
- So **we need a better method to Evaluate models.**
- That is **Cross Validation** method.

------------------------------------------------------

#### **2. What is K-Fold Cross Validation?**

- K-Fold Cross Validation is a technique where the dataset is divided into **K Equal parts**.

- Then the model is trained **K Times** using different train/test splits.

- Instead of testing the model **Once**, we test it multiple times.

------------------------------------------------------
**Example:**

Suppose we have **100 data points**

```py
we choose 
>> K=5

so we split data into 5 parts.
part 1
part 2
part 3
part 4
part 5
```
------------------------------------------------------
#### **Step-by-step Training**

**Iteration - 1**
```py
Train on:
part2 + part3 + part4 + part5

Test on:
part1
```
------------------------------------------------------
**Iteration - 2**
```py
Train on:
part1 + part3 + part4 + part5

Test on:
part2
```
------------------------------------------------------
**Iteration - 3**
```py
Train on:
part1 + part2 + part4 + part5

Test on:
part3
```
------------------------------------------------------
**Iteration - 4**
```py
Train on:
part1 + part2 + part3 + part5

Test on:
part4
```
------------------------------------------------------
**Iteration - 5**
```py
Train on:
part1 + part2 + part3 + part4

Test on:
part5
```
------------------------------------------------------
#### **Final Accuracy**

**Suppose we get:**

|    Fold  |  R^2 Score |
|----------|------------|
|    Fold1 |      0.78
|    Fold2 |      0.80
|    Fold3 |      0.76
|    Fold4 |      0.82
|    Fold5 |      0.79

**Final Accuracy:**
Average = (0.78 + 0.80 + 0.76 + 0.82 + 0.79)/5
Final Score = 0.79

- This is **more reliable Accuracy**.

#### **3. Why K-Fold Improves Accuracy**

**Important Point**
**K-Fold doesn't directly increases Accuracy.**

Instead it:
- Finds **Better Generalization**
- Prevents **Overfitting**
- Gives **Stable Performance**

Without KFold:
- Accuracy = random split

with KFold:
- Accuracy = average of many splits.

So the evaluation is **More Reliable.**

#### **5. What is Grid Search?**

Grid Search is used to find the **Best Hyperparameters** for a model.

Example Regression models have parameters like:

**Linear Regression**
- fit_intercept

**Ridge Regression**
- alpha

**Random Forest**
- n_estimators
- max_depth
- min_sample_split

- Instead of manually trying parameters, **Grid Search automatically tries all combinations**.
------------------------------------------------------
**Example**
```py
Suppose we want to tune:

alpha = [0.01, 0.1, 1, 10]

Grid Search will test:
alpha = 0.01
alpha = 0.1
alpha = 1
alpha = 10

- Then choose the best fit.
```
------------------------------------------------------

#### **Grid Search + Cross Validation**

- Grid Search usually uses **K-Fold internally**.

So process becomes:

```py
    For each parameter combination
                |
    Run K-Fold Cross Validation
                |
    Calculate Average Score
                |
    Select Best parameter
```
------------------------------------------------------
**Example Diagram**

```py
alpha = 0.01
    |-- KFold --> score = 0.72

alpha = 0.1
    |-- KFold --> score = 0.78

alpha = 1
    |-- KFold --> score = 0.81 <-- **Best**

alpha = 10
    |-- KFold --> score = 0.76
```
**Best Parameter**

- alpha = 1

------------------------------------------------------
#### **Python Example**
```py
from sklearn.model_selection import GridSearchCV
from sklearn.linear_model import Ridge

model = Ridge()

params = {
    "alpha": [0.01,0.1,1,10]
}

grid = GridSearchCV(
    model,
    param_grid=params,
    cv=5
)

grid.fit(X,y)

print(grid.best_params_)
print(grid.best_score_)
```
