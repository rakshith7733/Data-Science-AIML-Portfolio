**What is Machine Learning:**

- Machine learning is the way to teach computers how to learn from experience (data)/historical data and improve over time without explicit programming.

**General Understanding:** Machine learning is the process of training a machine programmatically to produce the desired outcome without manually programming every rule.

**General Example:**
- Asking a machine what the weather will be tomorrow.
- Asking the machine to identify whether the animal is a dog or a cat.
- Asking a machine to look at past customer salary data and predict their salary 10 years from now.
- House price prediction based on categories like area and number of rooms to find what the price of the house could be.


**Important in ML:**
- The ML model should have numeric input, as it can't take categorical columns; for example, flight booking: First class, economy class, business class. The machine identifies them as numeric/float/double/0&1 values.
- Data should not present any outliers.
- It should not have any Null values.

#### **Basic Concepts of Machine Learning**

1. **Data - The fuel of ML**
    - Just like a car needs fuel to run, Machine learning needs data to learn.

    Ex: 
- If we want to teach a computer to recongize dogs and cats we must provide.
    - Pictures of Dogs and Cats
    - Labels (saying this is a dog and this is a cat).


2. **Features - The important parts of Data**
- features are characterstics that helps the computer recognize things.

    Ex: 
- If we want to recognize fruits, features could be
    - Color (red, yellow, green)
    - Shape (round, long)
    - Size (Small, big)

    Ex: 
- For Spam email detection, Features could be:
    - Words in the email (Free money, Buy now)
    - Sender (Unknown sender vs Known contact)
    - Number of links in the email.

3. **Labels - The correct answers**
- Labels tell the ML model what the correct answer is during training

    Ex:
    - A picture of dog with the label 'Dog'.
    - A picture of cat with the label 'Cat'.

4. **Training - Learning from examples**
- Training is like studying for a test. The more examples a computer sees, the better it learns.

Ex: 
- If you see 100 pictures of apples and bananas, you will learn to tell the difference.

- For ML training means:
    - Showing the computer lots of examples.
    - Letting the computer find patterns.
    - Checking if the computer make correct predictions.

5. **Testing - Checking if the model learned well.**
- After training, we test the ML model with new data to see if it has learned properly.

Ex: 
- A teacher teaches a child about birds and then shows a new bird to check if the child recongnizes it.

6. **Predictions - Making Guesses Based on Learning.**
- Once trained, the ML model can predict new things.

Ex: 
- If an ML model has learned from thousands of dogs and cats pictures, it can look at a new picture and predict whether it's a dog or a cat.

### ```Types Of Machine learning```

- They are 3 main types of machine learning.

#### 1. **Supervised learning (learning with a teacher)**
- The computer learns from labeled data (where answers are given).

- Example:
    - teaching a baby what is an apple by showing different apples and saying 'This is an apple'.
    - The baby learns and can recognize new apples later.

##### **Where it is used?**
- Spam detection in emails.
- Face Recognition in phones
- Diseases Diagnosis.


#### 2. **Unsupervised Learning (learning without a teacher)**
- The computer does not have labels and finds patterns on its own.

- Example:
    - Imagine giving a child a box of mixed fruits without telling them the names.
    - the child may group them into categories without knowing their names.

##### **Where it is used?**
- **Customer Segmentation** in shopping apps (grouping users based on shopping habits).
- **Anomaly Detection** in banks (Finding fraud transactions).

##### 3. **Reinforcement Learning (Learning by Trail & Error/through experiences)**
- The computer learns by trail and error and gets rewards for correct actions.

- Example:
    - A baby learning to walk - If they fall, they try again until they get better.
    - A dog learning tricks - If it does a trick correctly, it gets a treat.

##### **Where it is used?**
- **Self-Driving Cars** (learning how to drive safetly).
- **Game AI** (AI Players learning to beat human players).
